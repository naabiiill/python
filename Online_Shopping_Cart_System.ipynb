```python
# ==========================================
# Shopping Cart Management System
# ==========================================

class ShoppingCart:

    # Constructor
    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name

        # Create an empty list if no items are provided
        if items is None:
            self.items = []
        else:
            self.items = items

    # Add an item
    def add_item(self, price):
        if price > 0:
            self.items.append(price)
            print(f"{price} added successfully.")
        else:
            print("Price must be positive.")

    # Remove an item
    def remove_item(self, price):
        if price in self.items:
            self.items.remove(price)
            print(f"{price} removed successfully.")
        else:
            print("Item not found.")

    # Calculate the total amount
    def calculate_total(self):
        total = sum(self.items)
        return total

    # Apply discount
    def apply_discount(self):
        total = self.calculate_total()

        if total >= 3000:
            total = total * 0.90

        return total

    # Display cart information
    def display_cart(self):

        print("\n================================")
        print("Customer Name:", self.customer_name)
        print("Items:", self.items)
        print("Number of Items:", len(self.items))
        print("Total Amount:", self.calculate_total())
        print("Final Amount:", self.apply_discount())
        print("================================")

    # Clear the cart
    def clear_cart(self):
        self.items.clear()
        print("Cart has been cleared.")

    # Operator overloading
    def __add__(self, other):

        new_name = self.customer_name + " & " + other.customer_name

        new_items = self.items + other.items

        return ShoppingCart(new_name, new_items)


# ==========================================
# Creating objects
# ==========================================

cart1 = ShoppingCart("Alice")
cart2 = ShoppingCart("Bob")

# Adding items
cart1.add_item(1000)
cart1.add_item(1500)
cart1.add_item(800)

cart2.add_item(500)
cart2.add_item(2000)
cart2.add_item(1200)

# Removing items
cart1.remove_item(800)
cart2.remove_item(500)

# Displaying carts
cart1.display_cart()
cart2.display_cart()

# Combining carts
combined_cart = cart1 + cart2

# Displaying combined cart
combined_cart.display_cart()
```

```python
# ==========================================
# Shopping Cart Management System
# ==========================================

class ShoppingCart:

    # Constructor
    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name

        # Create an empty list if no items are provided
        if items is None:
            self.items = []
        else:
            self.items = items

    # Add an item
    def add_item(self, price):
        if price > 0:
            self.items.append(price)
            print(f"{price} added successfully.")
        else:
            print("Price must be positive.")

    # Remove an item
    def remove_item(self, price):
        if price in self.items:
            self.items.remove(price)
            print(f"{price} removed successfully.")
        else:
            print("Item not found.")

    # Calculate the total amount
    def calculate_total(self):
        return sum(self.items)

    # Apply discount
    def apply_discount(self):
        total = self.calculate_total()

        if total >= 3000:
            total = total * 0.90

        return total

    # Display cart details
    def display_cart(self):

        product_names = {
            1000: "Keyboard",
            1500: "Mouse",
            2000: "Monitor",
            1200: "Headphones",
            800: "USB Drive",
            500: "Charger"
        }

        print("\n================================")
        print("Customer Name:", self.customer_name)

        print("\nItems in Cart:")

        for item in self.items:
            name = product_names.get(item, "Unknown Item")
            print(name, ":", item)

        print("\nNumber of Items:", len(self.items))
        print("Total Amount:", self.calculate_total())
        print("Final Amount:", self.apply_discount())
        print("================================")

    # Clear the cart
    def clear_cart(self):
        self.items.clear()
        print("Cart has been cleared.")

    # Overloading the + operator
    def __add__(self, other):

        new_name = self.customer_name + " & " + other.customer_name
        new_items = self.items + other.items

        return ShoppingCart(new_name, new_items)


# ==========================================
# Creating objects
# ==========================================

cart1 = ShoppingCart("Alice")
cart2 = ShoppingCart("Bob")

# Adding items to cart 1
cart1.add_item(1000)
cart1.add_item(1500)
cart1.add_item(800)

# Adding items to cart 2
cart2.add_item(500)
cart2.add_item(2000)
cart2.add_item(1200)

# Removing items
cart1.remove_item(800)
cart2.remove_item(500)

# Display individual carts
cart1.display_cart()
cart2.display_cart()

# Combine the carts
combined_cart = cart1 + cart2

# Display the combined cart
combined_cart.display_cart()

# Clear a cart (optional)
# cart1.clear_cart()
```
